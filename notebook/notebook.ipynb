```python
# data ingestion

from langchain_community.document_loaders import PyMuPDFLoader

loader = PyMuPDFLoader(
    file_path="../data/document.pdf"
)

pdf_content = loader.load()
print(pdf_content)
print(len(pdf_content))
```

```python
# data chunking
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
 chunk_size=200,
 chunk_overlap=50,
 separators=["\n\n", "\n", " ", ""]
)

# splitting the document
split_data = text_splitter.split_documents(pdf_content)
print(f"split {len(pdf_content)} documents into {len(split_data)} chunks")
```

```python
# import libs for emb and storage
import numpy as np
from sentence_transformers import SentenceTransformer
import chromadb
import uuid
from chromadb.config import Settings
from typing import List, Tuple, Dict, Any
from sklearn.metrics.pairwise import cosine_similarity
 
```

```python
# vector embedding
model = SentenceTransformer('all-MiniLM-L6-v2')
print(f"model loaded with dimentions: {model.get_embedding_dimension()}")

texts = [doc.page_content for doc in split_data]

# generate embedding
print(f"generating embedding for {len(texts)} texts")
embeddings = model.encode(texts)
print(f"generated embedding with shape: {embeddings.shape}")
```
